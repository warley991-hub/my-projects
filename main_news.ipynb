### Importando as bibliotecas

```python
from selenium import webdriver
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.common.by import By
from selenium.common.exceptions import NoSuchElementException, ElementClickInterceptedException
import time
```

### Iniciando o navegador e entrando no link

```python
navegador = webdriver.Chrome()

# Abrir o navegador no link
navegador.get('https://www.cnnbrasil.com.br/')
```

### Armazenando os elementos dos menus em variáveis

```python
menu_politica = navegador.find_element(By.XPATH,'/html/body/header/div/div/ul/li[3]/a')
menu_money = navegador.find_element(By.XPATH, '/html/body/header/div/div/ul/li[4]/a')
menu_infra = navegador.find_element(By.XPATH, '/html/body/header/div/div/ul/li[7]/a')
menu_ia = navegador.find_element(By.XPATH, '/html/body/header/div/div/ul/li[8]/a')
```

### Função de clique

```python
def clique(elemento):
    """
    Tenta clicar no elemento e fecha o pop-up se ele aparecer, independente do momento que ele apareça.

    Arg: 
        elemento --> O elemento web que será clicado.

    Erros:
        ElementClickInterceptedException: Caso o clique seja bloqueado por outro motivo que não seja o pop_up.
    """
    try:
        elemento.click()
    except ElementClickInterceptedException as e:
        try:
            pop_up = navegador.find_element('id','onesignal-slidedown-dialog')
            fechar_pop_up = navegador.find_element('id','onesignal-slidedown-cancel-button')
        except Exception:
            print(f'''
===============================
Erro: POP UP NÃO ENCONTRADO...
===============================''')
            raise e

        if pop_up:
            print(f'''
===================
POP UP ENCONTRADO!
===================
\n {pop_up}\n
===================
FECHANDO...
===================''')
            
        fechar_pop_up.click()
        
        print(f'''
==============================
POP UP FECHADO! CONTINUANDO...
==============================''')
        
        elemento.click()

    else:
        print(f'''
===================================
POP-UP NÃO ENCONTRADO...CONTINUANDO
===================================''')

```

### Função principal

```python
def main():           
    # Acessando o menu 'Política'
    time.sleep(3)
    clique(menu_politica)
    time.sleep(3)
    
    # Armazenando os elementos das notícias em dicionários
    acessar_noticia = {'noticia_1': navegador.find_element(By.ID, 'titulo-noticia-15456160')
                    }
    # Entrar na página da notícia
    clique(acessar_noticia['noticia_1'])
    
    # Armazenando a manchete
    manchete_noticia = {'noticia_1': navegador.find_element('tag name', 'h1').text
                         }
    # Armazenando o subtítulo
    subtitulo_noticia = {'noticia_1': navegador.find_element('tag name','h2').text
                        }
    print(f"""
==========
FUNCIONOU!
==========

MANCHETE:
{manchete_noticia['noticia_1']}

SUBTITULO:
{subtitulo_noticia['noticia_1']}
""")
     
main()
```
